Q1. Age Validation
Take age as input from the user. If the input is empty, raise a ValueError. If the age is negative, raise another ValueError. Use try-except to handle the errors. If the age is 18 or above, print "Eligible to vote"; otherwise, print "Not eligible to vote".

```python
try:
    age_input = input("Enter age: ")

    if age_input == "":
        raise ValueError("Age cannot be empty")

    age = int(age_input)

    if age < 0:
        raise ValueError("Age cannot be negative")

    if age >= 18:
        print("Eligible to Vote")

    else:
        print("Not eligible to vote")

except ValueError as e:
    print(e)
```

Q2. Positive Number Checker
Ask the user to enter a number. Handle invalid numeric input using try-except. If the number is zero or negative, raise a ValueError. If it is positive, check whether it is even or odd and display the result.

```python
num_input = int(input("Enter a number:"))

try:
    if num_input == 0 or num_input < 0:
        raise ValueError("Number can't be 0 or negative")
    elif num_input%2 == 0:
        print("It's and even number")
    else:
        print("It's and odd number")
except ValueError as e:
    print(e)
```

Q3. Username Validation
Take a username as input. If the username is empty or contains fewer than 5 characters, raise a ValueError. Use try-except to handle the exception. If the username is valid, print "Username accepted".

Q4. Division Calculator
Take two numbers as input. Handle invalid input using try-except. If the second number is zero, raise a ZeroDivisionError. Otherwise, divide the first number by the second and print the result. Use else to display the result only when no exception occurs.

```python
num1 = int(input("Enter num1"))
num2 = int(input("Enter num2"))
try:
    division = num1/num2
    if num2 == 0:
        raise ZeroDivisionError("Denominator can't be zero")
    else:
        print(division)
except ValueError as e:
    print(e)
```

Q5. Student Marks Validation
Take marks as input. If the input is not a valid integer, handle the exception. If marks are less than 0 or greater than 100, raise a ValueError. Otherwise, use if-else to print "Pass" if marks are at least 40 and "Fail" otherwise.

```python
marks = int(input("Enter marks"))
try: 
    if marks < 0 or marks > 100:
        raise ValueError("enter valid marks")
    elif marks >= 40:
        print("Pass")
    else:
        print("Fail")
except ValueError as e:
    print(e)
```

Q9. Read a File Safely
Ask the user to enter a filename. Open the file using "r" mode and display its contents. Handle FileNotFoundError if the file does not exist. If the file is empty, print "File is empty".

```python
file_name = "empty.txt"

try:
    with open(file_name, "r") as file:
        content = file.read()

        if content == "":
            raise ValueError("File is empty")

        print(content)

except ValueError as e:
    print(e)

except FileNotFoundError:
    print("File does not exist")
```

Q10. Number Storage in a File
Take five numbers from the user and write them to numbers.txt using "w" mode. If a user enters an invalid number, handle the exception. If a number is negative, raise a ValueError. Store only valid, non-negative numbers in the file.

```python
for i in range(1, 6): 
    try:
        num = int(input(f"Enter Number{i}:"))
        if num < 0:
            raise ValueError("Please enter a valid number.")
        file = open("numbers.txt", "a")
        file.write(f"{num} ")
        file.close()
    except ValueError as e:
        print(e)
    
```

Q11. Search for a Student
Read students.txt using "r" mode. Ask the user to enter a student name. If the name exists in the file, print "Student found"; otherwise, print "Student not found". Handle FileNotFoundError if the file does not exist.

```python
try: 
    with open("student.txt", "r") as file:  # Use with open(...) so Python closes the file automatically
        std_name = input("Enter student name:")
        content = file.read()
        if std_name in content:
            print("Student found")
        else:
            print("Student not found")
except FileNotFoundError:
    print("File not found")
```

Q13. Append Only Passing Students
Take a student's name and marks as input. If marks are outside the range 0–100, raise a ValueError. If marks are 40 or above, append the student's details to passed_students.txt using "a" mode. Otherwise, print "Student failed; record not added".

```python
try: 
    std_name = input("Enter name")
    if std_name == "":
        raise ValueError("Name can't be empty")
    std_marks = int(input("Enter marks"))
    if not 1 <= std_marks <= 100:
        raise ValueError("Marks not in range")
    if std_marks > 40:
        with open("passed_student.txt", "a") as file:
            file.write(f"Student Details:\nName:{std_name}, Marks:{std_marks}")
            print("Student Details added to file")
    else:
        print("Student failed")
except ValueError as e:
    print(e)
```

Q14. Count Lines in a File
Ask the user for a filename and read it using "r" mode. Count the number of non-empty lines in the file. Handle FileNotFoundError. If the file contains no non-empty lines, print "No records found".

```python
try:
    file_name = input("Enter file name")
    count = 0
    with open(file_name, "r") as file:
        for line in file:
            if line.strip() != "":
                count += 1
    if count == 0:
        print("No records found")
    else:
        print(count)
except FileNotFoundError:
    print("File not found")
```
