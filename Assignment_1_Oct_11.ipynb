```python
def func_ab_error(predicted_value):
    actual_value = 0.9
    absolute_error = abs(actual_value - predicted_value)
    return absolute_error

i = 1
while i<=10:
    predicted_value = float(input("Enter Predicted value ="))
    print("Absolute Error =",round(func_ab_error(predicted_value),))
    i = i + 1
```

```python
total_error = []
def func_mse(actual_value, predicted_value):
    squared_error = abs(actual_value - predicted_value) ** 2
    total_error.append(squared_error)
    
for i in range (10): 
    actual_value = float(input("Enter Actual values"))
    predicted_value = float(input("Enter Predicted values"))
    func_mse(actual_value, predicted_value)
    i = i + 1
mse = sum(total_error)/10    
print("Mean Squared Error:", mse)
```

```python
def calculate_mae(error1, error2, error3):
    abs1 = abs(error1)
    abs2 = abs(error2)
    abs3 = abs(error3)
    mae = (abs1+abs2+abs3)/3
    return mae

result = calculate_mae(0.2,-0.9,0.3)
print("Mean Absolute Error:",result)
```

```python
def func_final_loss(starting_loss,reduction_amount,epoch):
    for i in range(epoch):
        starting_loss = starting_loss - reduction_amount
    return starting_loss

final_loss = func_final_loss(3, 0.3, 3)
print("Final loss:",final_loss)
```

```python
def model_score(*score):
    count = len(score)
    highest_score = score[0]
    for i in range(1, count): 
        if score[i] > highest_score:
            highest_score = score[i]
    return highest_score,count

highest_score, count = model_score(5,6,8,3,7)
print("Highest score:", highest_score)
print("Number of arguments:", count)
```

```python
def validation_score(*score):
    count = len(score)
    total_score = sum(score)
    avg = total_score//count
    return avg

model_1 = validation_score(3,5,8,9,5,6)
print("Avg Model-1:",model_1)
model_2 = validation_score(8,6,2,3,1,0)
print("Avg Model-2:",model_2)
if model_1 > model_2:
    print("Model-1 has higher average")
else:
    print("Model-2 has higher average")
```
